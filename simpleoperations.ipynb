```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        fact = 1
        for i in range(1, n + 1):
            fact *= i
        return fact

# Example usage:
number = 7
print(f"The factorial of {number} is {factorial(number)}")
```

```python
# Function to display the multiplication table
def display_multiplication_table(num):
    print(f"Multiplication Table for {num}:")
    for i in range(1, 11): # Table from 1 to 10
        print(f"{num} x {i} = {num * i}")

# Example usage for the number 8
number_to_display = 8
display_multiplication_table(number_to_display)
```

```python
def check_odd_even(number):
    if number % 2 == 0:
        return f"{number} is an even number."
    else:
        return f"{number} is an odd number."

# Example usage:
input_number = 15
print(check_odd_even(input_number))
```

```python
def check_palindrome_number(number):
    # Convert the number to a string to easily reverse it
    num_str = str(number)
    # Reverse the string
    reversed_num_str = num_str[::-1]

    # Compare the original and reversed strings
    if num_str == reversed_num_str:
        return f"{number} is a palindrome number."
    else:
        return f"{number} is not a palindrome number."

# Example usage with 19:
input_number = 19
print(check_palindrome_number(input_number))

# Example with a palindrome number
input_number_2 = 121
print(check_palindrome_number(input_number_2))
```

## Represent Knowledge Base

### Subtask:
Implement the defined rules in a Python data structure, such as a list of dictionaries, where each dictionary represents a rule with conditions and conclusions.


**Reasoning**:
To implement the knowledge base as a list of dictionaries as instructed, I will create an empty list, define example rules as dictionaries with 'conditions' and 'conclusion' keys, append them to the list, and then print the list to verify the structure.



```python
rules = []

# Rule 1: IF has_feathers AND lays_eggs THEN is_bird
rule1 = {
    'conditions': ['has_feathers', 'lays_eggs'],
    'conclusion': 'is_bird'
}
rules.append(rule1)

# Rule 2: IF has_hair AND gives_milk THEN is_mammal
rule2 = {
    'conditions': ['has_hair', 'gives_milk'],
    'conclusion': 'is_mammal'
}
rules.append(rule2)

# Rule 3: IF is_bird AND can_fly THEN is_flying_bird
rule3 = {
    'conditions': ['is_bird', 'can_fly'],
    'conclusion': 'is_flying_bird'
}
rules.append(rule3)

# Print the rules list to verify its structure and content
print("Knowledge Base Rules:")
for i, rule in enumerate(rules):
    print(f"Rule {i+1}: {rule}")
```

## Implement Inference Engine

### Subtask:
Write a Python function that acts as the inference engine. This function will take user-provided facts (attributes) and use them to match against the rules in the knowledge base to deduce a decision or classification.


**Reasoning**:
I will define the `inference_engine` function as per the instructions, taking `facts` and `rules` as input, and implementing a forward-chaining mechanism to deduce new facts based on the knowledge base.



```python
def inference_engine(facts, rules):
    # Initialize an empty list to store facts deduced by the engine
    deduced_facts = []

    # This inference engine uses a simple forward-chaining mechanism.
    # It iterates through all rules and checks if their conditions are met by the known facts.
    # If all conditions are met, the rule's conclusion is added to the deduced facts.

    # Loop through each rule in the knowledge base
    for rule in rules:
        conditions = rule['conditions']
        conclusion = rule['conclusion']

        # Check if all conditions for the current rule are present in the known facts
        all_conditions_met = True
        for condition in conditions:
            if condition not in facts:
                all_conditions_met = False
                break # If one condition is not met, no need to check further for this rule

        # If all conditions are met, add the conclusion to the deduced facts
        if all_conditions_met:
            if conclusion not in deduced_facts: # Avoid adding duplicate conclusions
                deduced_facts.append(conclusion)

    return deduced_facts

print("Defined the inference_engine function.")
```
