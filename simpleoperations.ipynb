# Factorial

````python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        product = 1
        for i in range(1, n + 1):
            product *= i
        return product

print(factorial(5))
print(factorial(0))
print(factorial(-4))
````

# Odd & Even

````python
def check_odd_even(number):
    if number % 2 == 0:
        return f"{number} is an even number."
    else:
        return f"{number} is an odd number."

# Example usage:
input_number = 15
print(check_odd_even(input_number))
````

# Palindrome

````python
def check_palindrome_number(number):
    # Convert the number to a string to easily reverse it
    num_str = str(number)
    # Reverse the string
    reversed_num_str = num_str[::-1]

    # Compare the original and reversed strings
    if num_str == reversed_num_str:
        return f"{number} is a palindrome number."
    else:
        return f"{number} is not a palindrome number."

# Example usage with 19:
input_number = 19
print(check_palindrome_number(input_number))

# Example with a palindrome number
input_number_2 = 121
print(check_palindrome_number(input_number_2))
````

# Expert System

# Task
Develop an expert system capable of classifying animals based on a set of defined 'IF-THEN' rules. The system should include a knowledge base representing these rules, an inference engine to deduce classifications from given facts, and a command-line interface for user interaction. Finally, demonstrate and summarize its functionality.

## Define Decision Scenario and Rules

### Subtask:
Establish a simple decision-making scenario (e.g., animal classification) and define a set of 'IF-THEN' rules to represent the expert knowledge. For example, 'IF has_feathers AND can_fly THEN Bird'.


```markdown
### Decision Scenario: Animal Classification

We will create a simple expert system to classify animals based on a set of observable characteristics. This scenario is ideal for demonstrating IF-THEN rules because animal features are typically well-defined and can lead to clear classifications.

#### Key Facts/Attributes:
Here are the facts we will use to describe animals:
- `has_feathers`
- `lays_eggs`
- `can_fly`
- `has_hair`
- `gives_milk`
- `has_scales`
- `swims`

#### Expert Rules (IF-THEN):
These rules represent our expert knowledge for classifying animals:

1.  **Rule 1: Identifying Birds**
    -   **IF** `has_feathers` **AND** `lays_eggs`
    -   **THEN** `is_bird`

2.  **Rule 2: Identifying Mammals**
    -   **IF** `has_hair` **AND** `gives_milk`
    -   **THEN** `is_mammal`

3.  **Rule 3: Identifying Flying Birds (more specific bird classification)**
    -   **IF** `is_bird` **AND** `can_fly`
    -   **THEN** `is_flying_bird`

4.  **Rule 4: Identifying Fish**
    -   **IF** `has_scales` **AND** `swims`
    -   **THEN** `is_fish`

These rules will be translated into a programmatic knowledge base in the subsequent steps.
```

**Reasoning**:
The decision scenario and rules have been defined in the previous markdown block. The next logical step is to programmatically represent these rules and initial facts as Python data structures.



````python
possible_facts = [
    'has_feathers',
    'lays_eggs',
    'can_fly',
    'has_hair',
    'gives_milk',
    'has_scales',
    'swims'
]

rules = [
    # Rule 1: Identifying Birds
    {'conditions': ['has_feathers', 'lays_eggs'], 'conclusion': 'is_bird'},
    # Rule 2: Identifying Mammals
    {'conditions': ['has_hair', 'gives_milk'], 'conclusion': 'is_mammal'},
    # Rule 3: Identifying Flying Birds
    {'conditions': ['is_bird', 'can_fly'], 'conclusion': 'is_flying_bird'},
    # Rule 4: Identifying Fish
    {'conditions': ['has_scales', 'swims'], 'conclusion': 'is_fish'}
]

print("Defined possible facts:")
print(possible_facts)
print("\nDefined rules:")
for i, rule in enumerate(rules):
    print(f"Rule {i+1}: IF {', '.join(rule['conditions'])} THEN {rule['conclusion']}")
````

## Implement Inference Engine

### Subtask:
Write a Python function that acts as the inference engine. This function will take user-provided facts (attributes) and use them to match against the rules in the knowledge base to deduce a decision or classification.


**Reasoning**:
The subtask requires implementing an `inference_engine` function and demonstrating its usage with various examples. This code block will define the function as specified in the instructions and then apply it to several sets of initial facts to show different animal classifications.



````python
def inference_engine(initial_facts, rules):
    known_facts = set(initial_facts)

    while True:
        new_fact_deduced = False
        for rule in rules:
            # Check if all conditions for the rule are in known_facts
            all_conditions_met = all(condition in known_facts for condition in rule['conditions'])

            # If conditions are met and the conclusion is not already known
            if all_conditions_met and rule['conclusion'] not in known_facts:
                known_facts.add(rule['conclusion'])
                new_fact_deduced = True

        # If no new facts were deduced in this iteration, stop
        if not new_fact_deduced:
            break
    return known_facts

# Demonstration with different initial facts

# Example 1: A flying bird
facts1 = ['has_feathers', 'lays_eggs', 'can_fly']
deductions1 = inference_engine(facts1, rules)
print(f"\nInitial Facts for Example 1: {facts1}")
print(f"Deduced Facts for Example 1: {deductions1}")

# Example 2: A mammal
facts2 = ['has_hair', 'gives_milk']
deductions2 = inference_engine(facts2, rules)
print(f"\nInitial Facts for Example 2: {facts2}")
print(f"Deduced Facts for Example 2: {deductions2}")

# Example 3: A fish
facts3 = ['has_scales', 'swims']
deductions3 = inference_engine(facts3, rules)
print(f"\nInitial Facts for Example 3: {facts3}")
print(f"Deduced Facts for Example 3: {deductions3}")

# Example 4: A bird that doesn't fly (e.g., penguin, ostrich)
facts4 = ['has_feathers', 'lays_eggs']
deductions4 = inference_engine(facts4, rules)
print(f"\nInitial Facts for Example 4: {facts4}")
print(f"Deduced Facts for Example 4: {deductions4}")
````

## Implement Command-Line Interface (CLI) for User Interaction

### Subtask:
Develop a command-line interface that allows a user to input a list of facts about an animal. The interface should then use the previously defined `inference_engine` to deduce and display the animal's classification.

#### Instructions
1.  **Define a `query_expert_system` function**:
    *   This function will take `rules` and `possible_facts` as arguments.
    *   It should prompt the user to enter facts about an animal, separated by commas (e.g., `has_feathers, lays_eggs, can_fly`).
    *   Validate the user's input to ensure that only `possible_facts` are accepted. Provide feedback for invalid inputs.
    *   Call the `inference_engine` with the user's valid initial facts and the `rules`.
    *   Print the initial facts entered by the user.
    *   Print all the facts deduced by the inference engine, clearly separating the initial facts from the conclusions (e.g., 'is_bird', 'is_mammal').
    *   Summarize the classification (e.g., 'The animal is a flying bird.'). If multiple classifications are deduced, list them. If no specific classification (like 'is_bird', 'is_mammal', etc.) is found, state that the classification is inconclusive.
2.  **Demonstrate the `query_expert_system` function**:
    *   Call the `query_expert_system` function at least twice with different scenarios to showcase its functionality, including one where a clear classification is made and one where it might be inconclusive or unexpected.


**Reasoning**:
The previous steps defined the rules and the inference engine. Now, I need to implement the `query_expert_system` function as described in the subtask. This function will handle user input, validate facts, call the `inference_engine`, and present the deductions in a user-friendly format. I will also include demonstrations of this function.



````python
def query_expert_system(rules, possible_facts):
    print("\n--- Expert System Query ---")
    print(f"Enter facts about an animal, separated by commas. Possible facts are: {', '.join(possible_facts)}")
    user_input = input("Your facts (e.g., has_feathers, lays_eggs): ").strip().lower()

    input_facts_raw = [f.strip() for f in user_input.split(',') if f.strip()]

    valid_initial_facts = []
    invalid_facts = []

    for fact in input_facts_raw:
        if fact in possible_facts:
            valid_initial_facts.append(fact)
        else:
            invalid_facts.append(fact)

    if invalid_facts:
        print(f"Warning: The following entered facts are not recognized and will be ignored: {', '.join(invalid_facts)}")

    if not valid_initial_facts:
        print("No valid facts were entered. Cannot make a deduction.")
        return

    print(f"\nInitial valid facts: {', '.join(valid_initial_facts)}")

    # Use the inference engine to deduce new facts
    all_deduced_facts = inference_engine(valid_initial_facts, rules)

    # Separate initial facts from conclusions drawn by the engine
    deductions_only = sorted([fact for fact in all_deduced_facts if fact not in valid_initial_facts])

    print(f"All known facts after inference: {', '.join(sorted(list(all_deduced_facts)))}")

    if deductions_only:
        print(f"Inferred conclusions: {', '.join(deductions_only)}")
    else:
        print("No new conclusions were inferred.")

    # Summarize the classification
    classification = []
    if 'is_flying_bird' in all_deduced_facts:
        classification.append('a flying bird')
    elif 'is_bird' in all_deduced_facts:
        classification.append('a bird')
    if 'is_mammal' in all_deduced_facts:
        classification.append('a mammal')
    if 'is_fish' in all_deduced_facts:
        classification.append('a fish')

    if classification:
        print(f"Summary: The animal is likely {', and '.join(classification)}.")
    else:
        print("Summary: The classification is inconclusive based on the provided facts.")

# Demonstration of the query_expert_system function
print("\n--- Demonstration 1: A Flying Bird ---")
# Simulate user input for a flying bird: 'has_feathers, lays_eggs, can_fly'
# Note: In an actual Colab environment, this would require manual input or mock input handling.
# For automated execution, I'll provide a placeholder instruction.
print("Please enter 'has_feathers, lays_eggs, can_fly' when prompted.")
# query_expert_system(rules, possible_facts) # Uncomment to run interactively

print("\n--- Demonstration 2: A Mammal with an unrecognized fact ---")
print("Please enter 'has_hair, gives_milk, barks' when prompted.")
# query_expert_system(rules, possible_facts) # Uncomment to run interactively

print("\n--- Demonstration 3: Inconclusive ---")
print("Please enter 'swims' when prompted.")
# query_expert_system(rules, possible_facts) # Uncomment to run interactively

# To make the output visible in a non-interactive setting, let's call the function with hardcoded inputs
# for demonstration purposes in the notebook history.
# This is a workaround for the 'input()' function in a non-interactive execution environment.

# --- Hardcoded Demonstrations for notebook execution history ---
print("\n--- Hardcoded Demonstration 1: Flying Bird (for notebook history) ---")
# Simulating input 'has_feathers, lays_eggs, can_fly'
simulated_input_1 = 'has_feathers, lays_eggs, can_fly'
input_facts_raw_1 = [f.strip() for f in simulated_input_1.split(',') if f.strip()]
valid_initial_facts_1 = [fact for fact in input_facts_raw_1 if fact in possible_facts]
if valid_initial_facts_1:
    print(f"Initial valid facts: {', '.join(valid_initial_facts_1)}")
    all_deduced_facts_1 = inference_engine(valid_initial_facts_1, rules)
    deductions_only_1 = sorted([fact for fact in all_deduced_facts_1 if fact not in valid_initial_facts_1])
    print(f"All known facts after inference: {', '.join(sorted(list(all_deduced_facts_1)))}")
    if deductions_only_1:
        print(f"Inferred conclusions: {', '.join(deductions_only_1)}")
    else:
        print("No new conclusions were inferred.")
    classification_1 = []
    if 'is_flying_bird' in all_deduced_facts_1: classification_1.append('a flying bird')
    elif 'is_bird' in all_deduced_facts_1: classification_1.append('a bird')
    if 'is_mammal' in all_deduced_facts_1: classification_1.append('a mammal')
    if 'is_fish' in all_deduced_facts_1: classification_1.append('a fish')
    if classification_1: print(f"Summary: The animal is likely {', and '.join(classification_1)}.")
    else: print("Summary: The classification is inconclusive based on the provided facts.")

print("\n--- Hardcoded Demonstration 2: Mammal with unrecognized fact (for notebook history) ---")
# Simulating input 'has_hair, gives_milk, barks'
simulated_input_2 = 'has_hair, gives_milk, barks'
input_facts_raw_2 = [f.strip() for f in simulated_input_2.split(',') if f.strip()]
valid_initial_facts_2 = [fact for fact in input_facts_raw_2 if fact in possible_facts]
invalid_facts_2 = [fact for fact in input_facts_raw_2 if fact not in possible_facts]
if invalid_facts_2:
    print(f"Warning: The following entered facts are not recognized and will be ignored: {', '.join(invalid_facts_2)}")
if valid_initial_facts_2:
    print(f"Initial valid facts: {', '.join(valid_initial_facts_2)}")
    all_deduced_facts_2 = inference_engine(valid_initial_facts_2, rules)
    deductions_only_2 = sorted([fact for fact in all_deduced_facts_2 if fact not in valid_initial_facts_2])
    print(f"All known facts after inference: {', '.join(sorted(list(all_deduced_facts_2)))}")
    if deductions_only_2:
        print(f"Inferred conclusions: {', '.join(deductions_only_2)}")
    else:
        print("No new conclusions were inferred.")
    classification_2 = []
    if 'is_flying_bird' in all_deduced_facts_2: classification_2.append('a flying bird')
    elif 'is_bird' in all_deduced_facts_2: classification_2.append('a bird')
    if 'is_mammal' in all_deduced_facts_2: classification_2.append('a mammal')
    if 'is_fish' in all_deduced_facts_2: classification_2.append('a fish')
    if classification_2: print(f"Summary: The animal is likely {', and '.join(classification_2)}.")
    else: print("Summary: The classification is inconclusive based on the provided facts.")

print("\n--- Hardcoded Demonstration 3: Inconclusive (for notebook history) ---")
# Simulating input 'swims'
simulated_input_3 = 'swims'
input_facts_raw_3 = [f.strip() for f in simulated_input_3.split(',') if f.strip()]
valid_initial_facts_3 = [fact for fact in input_facts_raw_3 if fact in possible_facts]
invalid_facts_3 = [fact for fact in input_facts_raw_3 if fact not in possible_facts]
if invalid_facts_3:
    print(f"Warning: The following entered facts are not recognized and will be ignored: {', '.join(invalid_facts_3)}")
if valid_initial_facts_3:
    print(f"Initial valid facts: {', '.join(valid_initial_facts_3)}")
    all_deduced_facts_3 = inference_engine(valid_initial_facts_3, rules)
    deductions_only_3 = sorted([fact for fact in all_deduced_facts_3 if fact not in valid_initial_facts_3])
    print(f"All known facts after inference: {', '.join(sorted(list(all_deduced_facts_3)))}")
    if deductions_only_3:
        print(f"Inferred conclusions: {', '.join(deductions_only_3)}")
    else:
        print("No new conclusions were inferred.")
    classification_3 = []
    if 'is_flying_bird' in all_deduced_facts_3: classification_3.append('a flying bird')
    elif 'is_bird' in all_deduced_facts_3: classification_3.append('a bird')
    if 'is_mammal' in all_deduced_facts_3: classification_3.append('a mammal')
    if 'is_fish' in all_deduced_facts_3: classification_3.append('a fish')
    if classification_3: print(f"Summary: The animal is likely {', and '.join(classification_3)}.")
    else: print("Summary: The classification is inconclusive based on the provided facts.")
````
