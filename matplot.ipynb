# Normal Distribution

First, let's import the necessary libraries: `numpy` for numerical operations (specifically for generating random numbers from a normal distribution) and `matplotlib.pyplot` along with `seaborn` for plotting.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate 1000 random numbers from a normal distribution
# with mean (loc) = 0 and standard deviation (scale) = 1
# You can change these parameters to explore different distributions
mean = 0
std_dev = 1
num_samples = 1000
normal_data = np.random.normal(loc=mean, scale=std_dev, size=num_samples)

print(f"Generated {num_samples} samples from a normal distribution with mean={mean} and standard deviation={std_dev}.")
```

Now, let's visualize the generated data using a histogram. A histogram shows the distribution of a dataset by dividing the data into bins and counting how many data points fall into each bin.

```python
plt.figure(figsize=(10, 6))
sns.histplot(normal_data, kde=True, bins=30, color='skyblue')
plt.title('Histogram of a Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

# Mean,Median,Variance and Standard Deviation

Let's compute the mean, median, variance, and standard deviation of the `normal_data` generated earlier. These are fundamental statistical measures that describe the central tendency and spread of a dataset.

```python
mean_val = np.mean(normal_data)
median_val = np.median(normal_data)
variance_val = np.var(normal_data)
std_dev_val = np.std(normal_data)

print(f"Mean of the normal distribution: {mean_val:.4f}")
print(f"Median of the normal distribution: {median_val:.4f}")
print(f"Variance of the normal distribution: {variance_val:.4f}")
print(f"Standard Deviation of the normal distribution: {std_dev_val:.4f}")
```

# Probability

## Coin Toss Simulation and Probability

Let's simulate a coin toss experiment. We'll define the number of coin tosses, simulate the outcomes (Heads or Tails), and then compute the empirical probabilities.

```python
# Number of coin tosses to simulate
num_tosses = 1000

# Simulate coin tosses: 0 for Tails, 1 for Heads
# np.random.randint(0, 2, num_tosses) generates random integers between 0 (inclusive) and 2 (exclusive)
coin_toss_results = np.random.randint(0, 2, num_tosses)

# Count the number of heads (1s) and tails (0s)
heads_count = np.sum(coin_toss_results == 1)
tails_count = np.sum(coin_toss_results == 0)

# Calculate probabilities
probability_heads = heads_count / num_tosses
probability_tails = tails_count / num_tosses

print(f"Simulated {num_tosses} coin tosses.")
print(f"Number of Heads: {heads_count}")
print(f"Number of Tails: {tails_count}")
print(f"Probability of Heads: {probability_heads:.4f}")
print(f"Probability of Tails: {probability_tails:.4f}")
```

We can also visualize the distribution of outcomes using a bar plot.

```python
plt.figure(figsize=(6, 4))
sns.countplot(x=coin_toss_results, palette='viridis')
plt.title('Distribution of Coin Toss Outcomes')
plt.xlabel('Outcome (0: Tails, 1: Heads)')
plt.ylabel('Frequency')
plt.xticks([0, 1], ['Tails', 'Heads'])
plt.show()
```
